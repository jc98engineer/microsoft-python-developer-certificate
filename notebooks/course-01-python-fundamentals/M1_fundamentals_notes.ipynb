Addressing potential concerns and best practices

While loops and conditional statements are powerful tools, their overuse or improper implementation can lead to complex and difficult-to-maintain code. It's important to follow best practices to ensure code clarity and efficiency. 

Keep loops concise: Avoid nesting loops too deeply, as this can make the code harder to understand and debug. Break down complex loops into smaller, more manageable chunks. Think of it like organizing a closet: it's much easier to find what you're looking for when things are neatly arranged and categorized.

Use meaningful conditions: Ensure that the conditions used in conditional statements are clear and easy to understand. Avoid overly complex conditions that can obfuscate the code's logic. Clear and concise conditions make it easier for others (and your future self!) to understand the intent of your code.

Consider alternative constructs: In some cases, alternative constructs like list comprehensions or lambda functions can provide more concise and efficient ways to achieve the same result as loops and conditional statements. These constructs can sometimes provide a more elegant and readable solution, especially for simple operations.

Avoid infinite loops: Be cautious to ensure that the loop condition eventually becomes False. Failing to do so can cause an infinite loop, which might crash your program or require manual termination.

````python
# area de rectangulio 

def area_rectangle(width, height):
    """
    Calculate the area of a rectangle.
    Formula:
        area = width * height
    The function attempts to convert both inputs to floats. If either input
    cannot be converted to a numeric value, or if either value is less than
    than or equal to zero, the function returns None.
    Parameters
    ----------
    width : int | float | str
        Rectangle width. Must be convertible to a positive float.
    height : int | float | str
        Rectangle height. Must be convertible to a positive float.
    Returns
    -------
    float | None
        The rectangle area as a float if inputs are valid; otherwise, None.
    """
    try:
        width = float(width)
        height = float(height)
    except (ValueError, TypeError):
        return None

    if width <= 0 or height <= 0:
        return None

    area = width * height
    return area

````

````python
def calculate_diameter_circle(radius: int | float) -> float:
    """
    Calculate the diameter of a circle given its radius.

    Parameters
    ----------
    radius : int | float
        Radius of the circle. It must be a positive number.

    Returns
    -------
    float
        Diameter of the circle. Returns -1.0 if the radius is negative.
    """
    if radius < 0:
        return -1.0

    diameter = 2 * radius
    return float(diameter)
````

## Art of abstraction

- DRY (Don't Repeat Yourself) principle emerged in the 1990s, popularized by Andy Hunt and Dave Thomas in their seminal book, "The Pragmatic Programmer."
single source of truth, ensuring consistency and simplifying updates. 

- Now, what happens when we look at existing code and realize it's not as DRY or well-structured as it could be? That's where the idea of refactoring comes in.

- Refactoring is the process of restructuring existing computer code—changing the factoring—without changing its external behavior. Essentially, it's about cleaning up your code after it's written to improve its design, readability, and maintainability. Think of it as going back to those building blocks we talked about (functions) and rearranging or refining them to make the entire structure more solid and easier to work with in the long run.  

As a rule of thumb, consider these factors when deciding whether to create a function:

Frequency of use: If a code snippet is used multiple times, encapsulating it in a function promotes reusability and simplifies maintenance.

Complexity: If a task involves multiple steps or intricate logic, a function can improve readability and make the code more modular.

Likelihood of change: If a piece of code is likely to evolve over time, a function provides a centralized point of modification, reducing the risk of inconsistencies.

- Local: solo accesibles en ciertas partes del codigo
- Global: son accesible desde cualquier parte del codigo 
- Modifying a global variable in one function can have unintended consequences in other parts of your code that rely on it. This can lead to subtle bugs that are difficult to track down.

The LEGB rule: Python's search strategy for variables

When you use a variable in your code, Python embarks on a systematic search to find its value. The search order is described by the LEGB rule:

Local (L): Python first checks if the variable exists within the current function or code block.

Enclosing (E): If not found locally, it looks in enclosing functions (if you're using nested functions, like Russian dolls).

Global (G): Next, it searches the global scope, i.e., variables defined at the top level of your module.

Built-in (B): Finally, it checks Python's built-in namespace for pre-defined functions and objects like print, list, etc.

This hierarchical search ensures that Python always finds the most relevant value for a variable, preventing conflicts between variables with the same name at different levels.

````python
# Functional code block Temperature conversion tool
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

def fahrenheit_to_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5/9
    return celsius

def convert_temperature(temperature, unit):
    if unit == 'C':
        return celsius_to_fahrenheit(temperature)
    elif unit == 'F':
        return fahrenheit_to_celsius(temperature)
    else:
        return None
    
    
````

````python
# bubble sort 
# selection sort 
# Quicksort

def linear_search(data, target):
    """
    Perform a linear search to find the target element inside a list.

    Linear search checks each element one by one, from left to right,
    until it finds the target value or reaches the end of the list.

    Args:
        data (list): The list of elements to search through.
        target: The element we want to find.

    Returns:
        int | str:
            - The index of the target if it is found.
            - -1 if the target is not found.
            - An error message if data is not a list.
    """

    # 1. Validate that the input data is actually a list.
    # This prevents the function from failing if someone passes a string,
    # integer, dictionary, or another incorrect data type.
    if not isinstance(data, list):
        return "Invalid input: data must be a list"

    # 2. Handle the case of an empty list.
    # If the list has no elements, there is nothing to search.
    if not data:
        return -1

    # 3. Loop through the list using indexes.
    # range(len(data)) creates a sequence of indexes:
    # For example, if data has 5 elements, it creates: 0, 1, 2, 3, 4
    for i in range(len(data)):

        # 4. Compare the current element with the target.
        # data[i] means: "the element located at index i".
        if data[i] == target:

            # 5. If the current element matches the target,
            # return the index where the target was found.
            return i

    # 6. If the loop finishes and no match was found,
    # return -1 to indicate that the target is not inside the list.
    return -1
````


## Search Algorithms Comparison

| Algorithm | Requirement | Strategy | Speed |
|---|---|---|---|
| Linear Search | The list can be unordered | Checks each element one by one from left to right | Slower for large lists |
| Binary Search | The list must be sorted | Cuts the search space in half at each step | Much faster for large sorted lists |

## Main Search Algorithms

| Search Type | Main Idea | When It Is Useful |
|---|---|---|
| Linear Search | Checks each element one by one | Small datasets or unordered data |
| Binary Search | Repeatedly cuts the search space in half | Large datasets that are already sorted |
| Hash-Based Lookup | Uses a dictionary or set to check if a value exists | Very fast exact searches, such as checking IDs |
| Tree Search | Searches through hierarchical or indexed structures | Databases, indexes, file systems, and decision trees |
| Graph Search | Explores connected nodes | Routes, networks, dependencies, and recommendation systems |


````python
def binary_search(data, target):
    """
    Perform binary search to find the target element inside a sorted list.

    Binary search only works correctly when the input list is sorted.
    It repeatedly checks the middle element and eliminates half of the
    remaining search space.

    Args:
        data (list): A sorted list of elements.
        target: The element we want to find.

    Returns:
        int | str:
            - The index of the target if it is found.
            - -1 if the target is not found.
            - An error message if data is not a list.
    """

    # 1. Validate that data is a list.
    # This prevents errors if someone passes a string, number, dictionary, etc.
    if not isinstance(data, list):
        return "Invalid input: data must be a list"

    # 2. Define the search boundaries.
    # left represents the first index of the current search area.
    # right represents the last index of the current search area.
    left = 0
    right = len(data) - 1

    # 3. Continue searching while there is still a valid search area.
    # If left becomes greater than right, the target is not in the list.
    while left <= right:

        # 4. Find the middle index.
        # This avoids checking every element one by one.
        middle = (left + right) // 2 # floor division

        # 5. Get the value located at the middle index.
        middle_value = data[middle]

        # 6. If the middle value is the target, we found it.
        if middle_value == target:
            return middle

        # 7. If the target is greater than the middle value,
        # it must be on the right side of the list.
        elif target > middle_value:
            left = middle + 1

        # 8. If the target is smaller than the middle value,
        # it must be on the left side of the list.
        else:
            right = middle - 1

    # 9. If the loop finishes, the target was not found.
    return -1
````

````python
def quicksort(cards):
  if len(cards) < 2:
    return cards  # Base case: Already sorted if 0 or 1 element
  else:
    pivot = cards[0]  # Choose first card as pivot
    less = [i for i in cards[1:] if i <= pivot]
    greater = [i for i in cards[1:] if i > pivot]
    return quicksort(less) + [pivot] + quicksort(greater)
````
